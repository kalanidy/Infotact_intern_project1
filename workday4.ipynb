```python
#Day 3 – Standardize Timestamp
```

```python
#Convert timestamp into datetime.
events = pd.read_csv(r"C:\Users\yash6\Internship 1\Project 2\events.csv")
events["timestamp"] = pd.to_datetime(events["timestamp"])
```

```python
print(events["timestamp"].head())
```

```python
#Create new columns.
events["Date"] = events["timestamp"].dt.date
```

```python
events["Month"] = events["timestamp"].dt.month
```

```python
events["Year"] = events["timestamp"].dt.year
```

```python
events["Hour"] = events["timestamp"].dt.hour
```

```python
#Check duplicates
events.duplicated().sum()
```

```python
#Remove duplicates
events = events.drop_duplicates()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
events = pd.read_csv(r"C:\Users\yash6\Internship 1\Project 2\events.csv")
```

```python
#Total Sessions
events["session_id"].nunique()
```

```python
#Total Page Views
events["page_category"].value_counts()
```

```python
#Click-through Pattern
events["event_type"].value_counts()
```

```python
#Device Analysis
events["device_type"].value_counts()
```

```python
#Traffic Source
events["traffic_source"].value_counts()
```

```python
#Average Session Duration
events["session_duration_sec"].mean()
```

```python
#Chart 1
events["event_type"].value_counts().plot(kind="bar")
plt.title("Event Types")
plt.xlabel("Event Type")
plt.ylabel("Count")
plt.show()
```

```python
#Chart 2
events["traffic_source"].value_counts().plot(kind="pie",autopct="%1.1f%%")
plt.show()
```

```python
#Chart 3
events["session_duration_sec"].plot(kind="hist",bins=20)
plt.title("Session Duration")
plt.show()
```
