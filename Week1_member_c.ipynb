# Multi-Touch Marketing Attribution & ROI Dashboard

## Week 1 - Day 1

### Objective

Create the project data dictionary and understand the structure of all datasets before beginning preprocessing.

---

## Dataset Files

This project uses five relational datasets:

- customers.csv
- campaigns.csv
- events.csv
- products.csv
- transactions.csv

Together they represent the complete customer marketing journey from campaign exposure to purchase.

# Data Dictionary

## customers.csv

| Column | Description |
|---------|-------------|
| customer_id | Unique customer identifier |
| first_name | Customer first name |
| last_name | Customer last name |
| gender | Customer gender |
| age | Customer age |
| country | Customer country |
| city | Customer city |
| signup_date | Registration date |
| loyalty_tier | Customer loyalty level |

---

## campaigns.csv

| Column | Description |
|---------|-------------|
| campaign_id | Campaign identifier |
| campaign_name | Campaign name |
| channel | Marketing channel |
| campaign_type | Email, Social, Search etc. |
| start_date | Campaign start |
| end_date | Campaign end |

---

## events.csv

| Column | Description |
|---------|-------------|
| event_id | Event identifier |
| customer_id | Customer |
| session_id | Session |
| timestamp | Event time |
| event_type | View, Click, Purchase etc. |
| campaign_id | Campaign involved |
| product_id | Product |
| device_type | Mobile/Desktop |
| traffic_source | Organic/Paid |

---

## products.csv

| Column | Description |
|---------|-------------|
| product_id | Product identifier |
| product_name | Product |
| category | Category |
| brand | Brand |
| price | Selling price |

---

## transactions.csv

| Column | Description |
|---------|-------------|
| transaction_id | Purchase ID |
| customer_id | Customer |
| campaign_id | Campaign |
| product_id | Product |
| quantity | Quantity sold |
| revenue | Revenue |
| purchase_date | Purchase date |
