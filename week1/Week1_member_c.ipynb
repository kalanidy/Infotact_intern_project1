## CRM Data Cleaning

### Objective

Clean the CRM conversion data by:

- Identifying duplicate customer IDs
- Removing duplicate customer records
- Removing invalid customer records
- Removing test records
- Removing invalid transaction records
- Saving the cleaned CRM datasets

```python
customers.head()
```

```python
transactions.head()
```

## Check Dataset Size

Check the number of customer and transaction records before cleaning.

```python
print("Customers dataset:", customers.shape)

print("Transactions dataset:", transactions.shape)
```

## Identify Duplicate Customer IDs

The `customer_id` column should contain a unique identifier for every customer. Duplicate customer IDs are identified before removing them.

```python
duplicate_customers = customers.duplicated(
    subset="customer_id"
).sum()

print(
    "Duplicate customer IDs:",
    duplicate_customers
)
```

```python
customers_clean = customers.drop_duplicates(
    subset="customer_id",
    keep="first"
).copy()

print(
    "Customer records after deduplication:",
    customers_clean.shape
)
```

## Remove Invalid Customer Records

Customer records with missing customer IDs or invalid demographic values are removed.

```python
customers_clean = customers_clean.dropna(
    subset=["customer_id"]
)

customers_clean = customers_clean[
    customers_clean["age"].between(
        18,
        100
    )
]
```

## Remove Test Customer Records

Test records are identified using customer IDs containing values such as `test`, `dummy`, or `sample`.

```python
test_pattern = "test|dummy|sample"

test_customers = customers_clean[
    customers_clean["customer_id"]
    .astype(str)
    .str.contains(
        test_pattern,
        case=False,
        na=False
    )
]

print(
    "Test customer records:",
    len(test_customers)
)
```

```python
customers_clean = customers_clean[
    ~customers_clean["customer_id"]
    .astype(str)
    .str.contains(
        test_pattern,
        case=False,
        na=False
    )
]
```

## Clean Transaction Records

Remove:

- Duplicate transaction IDs
- Missing transaction IDs
- Missing customer IDs
- Invalid quantities
- Negative revenue values

```python
transactions_clean = (
    transactions
    .drop_duplicates(
        subset="transaction_id",
        keep="first"
    )
    .dropna(
        subset=[
            "transaction_id",
            "customer_id"
        ]
    )
    .copy()
)
```

```python
transactions_clean = transactions_clean[
    (transactions_clean["quantity"] > 0)
    &
    (transactions_clean["gross_revenue"] >= 0)
]
```

```python
valid_customer_ids = set(
    customers_clean["customer_id"]
)

transactions_clean = transactions_clean[
    transactions_clean["customer_id"]
    .isin(valid_customer_ids)
]
```

## Validate Cleaned CRM Data

Verify that duplicate customer IDs and invalid records have been removed successfully.

```python
print(
    "Duplicate customer IDs:",
    customers_clean
    .duplicated(
        subset="customer_id"
    )
    .sum()
)

print(
    "Duplicate transaction IDs:",
    transactions_clean
    .duplicated(
        subset="transaction_id"
    )
    .sum()
)

print(
    "Missing customer IDs:",
    customers_clean["customer_id"]
    .isnull()
    .sum()
)

print(
    "Missing transaction IDs:",
    transactions_clean["transaction_id"]
    .isnull()
    .sum()
)
```

```python
print("CRM Cleaning Summary")

print(
    "Customers before:",
    len(customers)
)

print(
    "Customers after:",
    len(customers_clean)
)

print(
    "Transactions before:",
    len(transactions)
)

print(
    "Transactions after:",
    len(transactions_clean)
)
```

## Save Cleaned CRM Data

The cleaned customer and transaction datasets are saved for future EDA, SQL analysis, and attribution modeling.

```python
customers_clean.to_csv(
    "../customers_clean.csv",
    index=False
)

transactions_clean.to_csv(
    "../transactions_clean.csv",
    index=False
)

print(
    "Cleaned CRM datasets saved successfully."
)
```

## Exploratory Data Analysis on CRM Conversions

### Objective

Perform exploratory data analysis on CRM conversion data to understand:

- Conversion volume
- Conversion timing
- Conversion value

The analysis uses the cleaned customer and transaction datasets created on Day 3.

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
customers_clean = pd.read_csv(r"A:\Multi-Touch Marketing Attribution\customers_clean.csv"
)

transactions_clean = pd.read_csv(r"A:\Multi-Touch Marketing Attribution\transactions_clean.csv"
)
```

```python
transactions_clean["timestamp"] = pd.to_datetime(
    transactions_clean["timestamp"]
)
```

## 1. Conversion Volume Analysis

Analyze the total number of purchase transactions and how conversion volume changes over time.

```python
total_conversions = len(transactions_clean)

print("Total Conversions:", total_conversions)
```

```python
daily_conversions = (
    transactions_clean
    .groupby(
        transactions_clean["timestamp"].dt.date
    )
    .size()
)

daily_conversions.head()
```

```python
plt.figure(figsize=(12, 5))

daily_conversions.plot()

plt.title("Daily CRM Conversion Volume")
plt.xlabel("Date")
plt.ylabel("Number of Conversions")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

### Observation

The daily conversion trend shows how purchase activity changes over time. Peaks indicate periods of higher customer conversion activity, while lower points represent periods with fewer purchases.

## 2. Conversion Timing Analysis

Analyze when customers are most likely to complete purchases by examining conversion activity by hour of the day.

```python
transactions_clean["conversion_hour"] = (
    transactions_clean["timestamp"].dt.hour
)
```

```python
hourly_conversions = (
    transactions_clean
    .groupby("conversion_hour")
    .size()
)

hourly_conversions
```

```python
plt.figure(figsize=(10, 5))

hourly_conversions.plot(kind="bar")

plt.title("Conversions by Hour of Day")
plt.xlabel("Hour of Day")
plt.ylabel("Number of Conversions")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
peak_hour = hourly_conversions.idxmax()
peak_conversions = hourly_conversions.max()

print("Peak Conversion Hour:", peak_hour)
print("Conversions During Peak Hour:", peak_conversions)
```

### Observation

The hourly conversion analysis helps identify the time of day with the highest purchase activity. This information can support campaign scheduling and marketing decisions.

## 3. Conversion Value Analysis

Analyze the monetary value generated by customer purchases using the `gross_revenue` column.

```python
total_revenue = transactions_clean["gross_revenue"].sum()

average_conversion_value = (
    transactions_clean["gross_revenue"].mean()
)

median_conversion_value = (
    transactions_clean["gross_revenue"].median()
)

print(f"Total Revenue: ${total_revenue:,.2f}")
print(
    f"Average Conversion Value: "
    f"${average_conversion_value:,.2f}"
)
print(
    f"Median Conversion Value: "
    f"${median_conversion_value:,.2f}"
)
```

```python
plt.figure(figsize=(10, 5))

plt.hist(
    transactions_clean["gross_revenue"],
    bins=30,
    edgecolor="black"
)

plt.title("Distribution of Conversion Value")
plt.xlabel("Gross Revenue")
plt.ylabel("Number of Transactions")
plt.tight_layout()
plt.show()
```

### Observation

The conversion value distribution shows the frequency of low-value and high-value transactions. Comparing the average and median conversion values helps identify whether unusually large purchases influence the overall revenue distribution.

## 4. Revenue Trend Over Time

## CRM and Web Analytics Validation

### Objective

Validate the cleaned CRM conversion data against Web Analytics data by matching customer IDs across:

- Customer records
- Transaction records
- Web Analytics events

This ensures that CRM customers and conversions can be connected to their corresponding web interactions.

```python
print("Total Customers:", len(customers_clean))
print("Total Transactions:", len(transactions_clean))
print("Total Web Events:", len(events))
```

## 1. Validate Customer IDs Between CRM and Web Analytics

Extract unique customer IDs from both datasets and identify how many CRM customers are present in the Web Analytics data.

```python
crm_customer_ids = set(
    customers_clean["customer_id"].dropna().unique()
)

web_customer_ids = set(
    events["customer_id"].dropna().unique()
)
```

```python
matched_customer_ids = (
    crm_customer_ids.intersection(web_customer_ids)
)

crm_only_ids = (
    crm_customer_ids.difference(web_customer_ids)
)

web_only_ids = (
    web_customer_ids.difference(crm_customer_ids)
)
```

```python
print("Unique CRM Customers:", len(crm_customer_ids))
print("Unique Web Customers:", len(web_customer_ids))

print(
    "Matched Customer IDs:",
    len(matched_customer_ids)
)

print(
    "CRM Customers Not Found in Web Analytics:",
    len(crm_only_ids)
)

print(
    "Web Customers Not Found in CRM:",
    len(web_only_ids)
)
```

```python
match_percentage = (
    len(matched_customer_ids)
    / len(crm_customer_ids)
    * 100
)

print(
    f"CRM to Web Analytics Match Rate: "
    f"{match_percentage:.2f}%"
)
```

## 2. Validate Transaction Customers Against Web Analytics

Check whether customers who completed purchases also have corresponding Web Analytics activity.

```python
transaction_customer_ids = set(
    transactions_clean["customer_id"]
    .dropna()
    .unique()
)
```

```python
matched_transaction_customers = (
    transaction_customer_ids
    .intersection(web_customer_ids)
)

unmatched_transaction_customers = (
    transaction_customer_ids
    .difference(web_customer_ids)
)
```

```python
print(
    "Unique Customers with Transactions:",
    len(transaction_customer_ids)
)

print(
    "Transaction Customers Found in Web Analytics:",
    len(matched_transaction_customers)
)

print(
    "Transaction Customers Not Found in Web Analytics:",
    len(unmatched_transaction_customers)
)
```

```python
transaction_match_rate = (
    len(matched_transaction_customers)
    / len(transaction_customer_ids)
    * 100
)

print(
    f"Transaction Customer Match Rate: "
    f"{transaction_match_rate:.2f}%"
)
```

## 3. Validate Individual Transactions Against Web Analytics

Add a validation flag to each transaction indicating whether the customer appears in the Web Analytics dataset.

```python
transactions_validated = transactions_clean.copy()

transactions_validated["web_analytics_match"] = (
    transactions_validated["customer_id"]
    .isin(web_customer_ids)
)
```

```python
transactions_validated[
    "web_analytics_match"
].value_counts()
```

```python
unmatched_customers = pd.DataFrame({
    "customer_id": list(
        unmatched_transaction_customers
    )
})

unmatched_customers.head()
```

```python
transactions_validated.to_csv(
    "../transactions_validated.csv",
    index=False
)

print(
    "Validated transaction data saved successfully."
)
```
